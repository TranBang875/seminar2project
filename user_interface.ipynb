```python
import gradio as gr
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
```

```python
MODEL_NAME = "/content/drive/MyDrive/seminar 2/checkpoint-7979"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSequenceClassification.from_pretrained(MODEL_NAME)

labels = ["toxic", "severe_toxic", "obscene", "threat", "insult", "identity_hate"]
```

```python
def classify_comment(text):
    if not text.strip():
        return {label: 0.0 for label in labels}

    inputs = tokenizer(text, return_tensors="pt", truncation=True, padding=True)
    with torch.no_grad():
        outputs = model(**inputs)
        probs = torch.sigmoid(outputs.logits).squeeze().tolist()

    result = {label: float(f"{p:.3f}") for label, p in zip(labels, probs)}
    return result
```

```python
demo = gr.Interface(
    fn=classify_comment,
    inputs=gr.Textbox(
        lines=4,
        placeholder="Nhập comment tại đây...",
        label="Comment Input",
    ),
    outputs=gr.Label(num_top_classes=6, label="Toxicity Prediction"),
    title="🧠 Toxic Comment Classifier",
    description="Categorize the toxicity level of comments",
    examples=[
        ["You are a wonderful person!"],
        ["I hate you, you idiot!"],
        ["Go die, loser!"],
        ["Nice work team, great job!"]
    ],
)
```

```python
demo.launch(share=True)
```
